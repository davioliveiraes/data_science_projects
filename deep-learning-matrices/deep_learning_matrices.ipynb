```python
# imports
import numpy as np
```

### Parte 1 - Escrever o Algoritmo

```python
class AlgoritmoNeuralNetworkDSA:
   def __init__(self, taxa_aprendizado, num_interacoes):
      self.taxa_aprendizado = taxa_aprendizado
      self.num_interacoes = num_interacoes

      self.pesos = None
      self.bias = None
   
   def func_activation_sigmoid(self, pred):
      sig = 1 / (1 + np.exp(-pred))

      return sig
   
   def fit(self, X, y):
      num_registros, num_atributos = X.shape
      self.pesos = np.zeros(num_atributos)
      self.bias = 0
      print('\nTreinamento Inciado!')
   
      for i in range(self.num_interacoes):
         print("\nTreinamento do Modelo na Iteração", i)
         previsao = np.dot(X, self.pesos) + self.bias
         print(f"Previção antes da função de ativação: {previsao}")

         previsao_final = self.func_activation_sigmoid(previsao)
         print(f"Previsão depois da função de ativação: {previsao_final}")

         erro = (previsao_final - y)
         print(f"Erro do modelo: {erro}")

         dw = (1 / num_registros) * np.dot(X.T, erro)
         db = (1 / num_registros) * np.sum(previsao_final - y)

         self.pesos -= self.taxa_aprendizado * dw
         print(f"Valores de pesos: {self.pesos}")

         self.bias = self.taxa_aprendizado * db
         print(f"Valor de bias: {self.bias}")
      
      print("\nTreinamento Concluído!")
   
   def predict(self, X):
      previsao = np.dot(X, self.pesos) + self.bias
      print(f"Previsão depois de passar pela função de ativação: {previsao}")

      previsao_final = self.func_activation_sigmoid(previsao)
      print(f"Previsão depois de passar pela função de ativação: {previsao_final}")

      classe_prevista = [1 if i > 0.5 else 0 for i in previsao_final]
      
      return classe_prevista
```

> Vamos user 6 registros para treinar o modelo e 2 para avaliar o modelo depois de treinado.

```python
X_treino = np.array([[1, 2.5], [2, 3], [3, 5], [1, 4], [5, 6], [6, 7]])
y_treino = np.array([0, 0, 1, 0, 1, 1])
```

### Parte 3 - Treinamento do Modelo

```python
taxa_aprendizado = 0.01
num_itaracoes = 1000
```

```python
modelo_dsa = AlgoritmoNeuralNetworkDSA(taxa_aprendizado, num_itaracoes)
```

```python
modelo_dsa.fit(X_treino, y_treino)
```

### Parte 4 - Avaliação do Modelo

```python
X_test = np.array([[1.5, 2],[4, 5.5]])
y_test = np.array([0, 1])
```

```python
previsoes_teste = modelo_dsa.predict(X_test)
```

```python
for i, previsao in enumerate(previsoes_teste):

   entrada = X_test[i]
   saida = y_test[i]

   if previsao == 0:
      print(f"\nEntrada {entrada} e saída {saida} a previsão foi {previsao}: Não é Transação Suspeita")
   else:
      print(f"\nEntrada {entrada} e saída {saida} a previsão foi {previsao}: É Transação Suspeita")

print("\n")
```
