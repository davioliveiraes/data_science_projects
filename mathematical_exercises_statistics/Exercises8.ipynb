## Exercício 1

Vamos imaginar um cenário de negócios onde uma empresa de consultoria realiza análises de mercado para cinco diferentes setores: tecnologia, saúde, finanças, energia e varejo. A empresa tem um modelo de negócios no qual, a cada mês, escolhe aleatoriamente um desses setores para oferecer uma análise de mercado gratuita como parte de sua estratégia de marketing para atrair novos clientes. Suponha que a escolha do setor seja feita de forma que cada um tenha igual probabilidade de ser selecionado. 

Calcule a probabilidade de cada setor ser escolhido ao longo de um ano.

```python
setores = ['Tecnologias', 'Saúde', 'Finanças', 'Energia', 'Varejo']
total_setores = len(setores)
meses_ano = 12

print(" === ANÁLISE DE PROBABILIDADE === ")
print(f"Setores disponíveis: {total_setores}")
print(f"Meses do ano: {meses_ano}")

prob_mensal = 1 / total_setores
print("\nProbabilidade mensal por setor:")
print(f"P(setor) = 1/{total_setores} = {prob_mensal:.4f} ou {prob_mensal*100:.1f}%")

print("\nProbabilidade de cada setor ao longo do ano: ")

for setor in setores:
   prob_nunca = (1 - prob_mensal) ** meses_ano
   prob_pelo_menos_uma = 1 - prob_nunca

   print(f"{setor:12}: {prob_pelo_menos_uma:.4f} ({prob_pelo_menos_uma*100:.1f}%)")

import random
import numpy as np
import math 

def simular_ano():
   """Simula um ano de esolhas aleatórias"""
   escolhas = []
   for i in range(12):
      setor_escolhido = random.choice(setores)
      escolhas.append(setor_escolhido)
   return escolhas

print("\nSIMULAÇÃO (10.000 ANOS):")
simulacoes = 10000
resultados = {setor: 0 for setor in setores}

for _ in range(simulacoes):
   ano_simulado = simular_ano()
   setores_no_ano = set(ano_simulado)
   for setor in setores_no_ano:
      resultados[setor] += 1

print("Probabilidade observada na simulação:")
for setor in setores:
   prob_simulada = resultados[setor] / simulacoes
   print(f"{setor:12}: {prob_simulada:.4f} ({prob_simulada*100:.1f}%)")

print("\nESTATÍSTICAS EXTRAS:")
print(f"Números esperado de vezes que cada setor aparece por ano: {meses_ano}")
print(f"Probabilidade de um setor específico aparecer exatamente 3 vezes: {np.exp(-meses_ano*prob_mensal) * (meses_ano*prob_mensal)**3 / math.factorial(3):.2f}")

print("\nRESPOSTA:")
print(f"Cada setor tem {prob_pelo_menos_uma:.1f}% de chance de ser escolhido pelo menos uma vez durante o ano")
```

```python
# Outra solução
# Definindo os parâmetros do problema
numero_de_setores = 5
meses_no_ano = 12

# Cada setor tem uma igual probabilidade de ser escolhido a cada mês
# Então a probabilidade de cada setor ser escolhido em um mês específico é 1/numero_de_setores
# Ao longo de um ano, essa probabilidade não muda, pois cada mês é um evento independente.

# Calculando a probabilidade
probabilidade_por_mes = 1 / numero_de_setores

# Como queremos a probabilidade ao longo do ano, mas cada mês é independente
# E estamos interessados apenas na probabilidade de escolha mensal (não acumulativa ao longo do ano)
# A probabilidade de um setor ser escolhido em qualquer mês específico permanece a mesma.

probabilidade_ao_longo_do_ano = probabilidade_por_mes
print(f"Probabilidade ao longo do ano: {probabilidade_ao_longo_do_ano*100:.1f}%")
```

## Exercício 2

Considere uma empresa que está lançando uma nova campanha de marketing por e-mail, destinada a promover um de seus produtos. A partir de experiências passadas, a empresa sabe que a taxa de abertura esperada para esses e-mails é de 20%. A campanha envolve o envio de 100 e-mails para potenciais clientes. 

A empresa está interessada em saber qual é a probabilidade de pelo menos 25 desses e-mails serem abertos, o que considera um indicador de sucesso da campanha.

```python
import numpy as np
from scipy import stats # type: ignore
import math

n = 100
p = 0.20
sucesso_minimo = 25

print(" === ANÁLISE DE CAMPANHA DE E-MAIL === ")
print(f"E-mails enviados: {n}")
print(f"Meta de sucesso: pelo menos {sucesso_minimo} aberturas")

media = n * p
variancia = n * p * (1 - p)
desvio_padrao = math.sqrt(variancia)

print("PARÂMETROS DA DISTRIBUIÇÃO:")
print(f"Média esperada: {media:.1f} aberturas")
print(f"Desvio padrão: {desvio_padrao:.2f}")
print(f"Variância: {variancia:.1f}")

print("\n MÉTODO 1: DISTRIBUIÇÃO BINOMIAL EXATA")

prob_ate_24 = stats.binom.cdf(24, n, p)
prob_pelo_menos_25 = 1 - prob_ate_24

print(f"P(X <= 24) = {prob_ate_24:.6f}")
print(f"P(X >= 25) = 1 - P(X <= 24) = {prob_pelo_menos_25:.6f}")
print(f"Probabilidade de sucesso: {prob_pelo_menos_25*100:.2f}%")

print(f"MÉTODO 2: APROXIMAÇÃO NORMAL")

z_score = (24.5 - media) / desvio_padrao
prob_normal = 1 - stats.norm.cdf(z_score)

print(f"Z-score = (24.5 - {media:.1f}) / {desvio_padrao:.2f} = {z_score:.3f}")
print(f"P(X >= 25) ≈ {prob_normal:.6f}")
print(f"Aproximação normal: {prob_normal*100:.2f}%")

print("SIMULAÇÃO (10.000 campanhas):")
simulacoes = 10000
sucessos = 0

for _ in range(simulacoes):
   aberturas = np.random.binomial(n, p)
   if aberturas >= sucesso_minimo:
      sucessos += 1

prob_simulada = sucessos / simulacoes
print(f"Campanhas simuladas: {simulacoes}")
print(f"Sucessos observados: {sucessos}")
print(f"Probabilidades simulada: {prob_simulada:.4f} ({prob_simulada*100:.2f}%)")

print("PROBABILIDADES ADICIONAIS:")

prob_exata_20 = stats.binom.pmf(20, n , p)
print(f"P(exatamente 20 aberturas) = {prob_exata_20:.4f} ({prob_exata_20*100:.2f}%)")

prob_15_25 = stats.binom.cdf(25, n, p) - stats.binom.cdf(14, n, p)
print(f"P(15 <= X <= 25) = {prob_15_25:.4f} ({prob_15_25*100:.2f}%)")

intervalo_inf = stats.binom.ppf(0.025, n, p)
intervalo_sup = stats.binom.ppf(0.975, n, p)
print(f"Intervalo 95% de confiança: [{intervalo_inf:.0f}, {intervalo_sup:.0f}] aberturas")

print("INTERPRATAÇÃO:")
if prob_pelo_menos_25 > 0.5:
   print(f"PROVÁVEL: {prob_pelo_menos_25*100:.1f}% de chance de sucesso")
elif prob_pelo_menos_25 > 0.1:
   print(f"POSSÍVEL: {prob_pelo_menos_25*100:.1f}% de chance de sucesso")
else:
   print(f"IMPROVÁVEL: Apenas {prob_pelo_menos_25*100:.1f}% de chance de sucesso")

print(f"\nA empresa tem um chance de {prob_pelo_menos_25*100:.2f}% de atingir sua meta de pelo menos 25 aberturas.")
```

```python
# outra solução
from scipy.stats import binom

# Define os parâmetros da distribuição binomial
n = 100 # número total de e-mails enviados
p = 0.2 # probabilidade d abertura de cada e-mail

# Calcula a probabilidade de obter até 24 aberturas de e-mail
probabilidade_ate_24 = binom.cdf(24, n, p)

# Calcula a probabilidade de obter pelo menos 25 aberturas de e-mail
probabilidade_pelo_menos_25 = 1 - probabilidade_ate_24

print(f"A probabilidade de pelo menos 25 e-mails serem abertos é de {probabilidade_pelo_menos_25 * 100:.2f}%")
```

## Exercício 3

Imagine uma loja de varejo online que, historicamente, recebe em média 12 pedidos por hora. A gestão deseja entender melhor como isso se traduz em termos de fluxo de pedidos ao longo do dia para otimizar suas operações, incluindo alocação de pessoal, gestão de estoque e capacidade de atendimento ao cliente.

Qual a probabilidade da loja receber exatamente 15 pedidos em uma determinada hora?

```python
import numpy as np
from scipy import stats
import math
import matplotlib.pyplot as plt

lambda_pedidos = 12
pedidos_desajados = 15

print(" === ANÁLISE DE FLUXO DE PEDIDOS ONLINE === ")
print(f"Média histórica: {lambda_pedidos} pedidos/hora")
print(f"Pergunta: P(exatamente {pedidos_desajados} pedidos em 1 hora)")

print("\nCARACTERÍSTICAS DA DISTRIBUIÇÃO POISSON:")
print(f"Parâmetro λ = {lambda_pedidos}")
print(f"Média = {lambda_pedidos}")
print(f"Variância = {lambda_pedidos}")
print(f"Desvio padrão = {math.sqrt(lambda_pedidos):.2f}")

print("\nCÁLCULO PRINCIPAL:")

prob_exato_15 = stats.poisson.pmf(pedidos_desajados, lambda_pedidos)
print(f"P(X = {pedidos_desajados}) = (e(-{lambda_pedidos}) x {lambda_pedidos}^{pedidos_desajados}) / {pedidos_desajados}!")
print(f"P(X = {pedidos_desajados}) = {prob_exato_15:.6f}")
print(f"Probabilidade: {prob_exato_15*100:.2f}%")

e_neg_lambda = math.exp(-lambda_pedidos)
lambda_k = lambda_pedidos ** pedidos_desajados
k_factorial = math.factorial(pedidos_desajados)
prob_manual = (e_neg_lambda * lambda_k) / k_factorial

print("VERIFICAÇÃO MANUAL:")
print(f"e(-{lambda_pedidos}) = {e_neg_lambda:.6f}")
print(f"{lambda_pedidos}^{pedidos_desajados} = {lambda_k}")
print(f"{pedidos_desajados}! = {k_factorial}")
print(f"Resultado manual: {prob_manual:.6f}")

print("\n SIMULAÇÃO (10.000 horas):")
simulacoes = 10000
contador_15_pedidos = 0

pedidos_simulados = np.random.poisson(lambda_pedidos, simulacoes)
contador_15_pedidos = np.sum(pedidos_simulados == pedidos_desajados)

prob_simulada = contador_15_pedidos / simulacoes
print(f"Horas simuladas: {simulacoes}")
print(f"Horas com exatamente 15 pedidos: {contador_15_pedidos}")
print(f"Probabilidade simulada: {prob_simulada:.4f} ({prob_simulada*100:.2f}%)")

print("\n ANÁLISE COMPLEMENTARES PARA GESTÃO:")

prob_10_15 = stats.poisson.cdf(15, lambda_pedidos) - stats.poisson.cdf(9, lambda_pedidos)
print(f"P(10 <= X <= 15) = {prob_10_15:.4f} ({prob_10_15*100:.1f}%)")

prob_mais_15 = 1 - stats.poisson.cdf(15, lambda_pedidos)
print(f"P(X > 15) = {prob_mais_15:.4f} ({prob_mais_15*100:.1f}) - Risco de sobrecarga")

prob_menos_10 = stats.poisson.cdf(9, lambda_pedidos)
print(f"P(X < 10) = {prob_menos_10:.4f} ({prob_menos_10*100:.1f}%) - Período calmo")

intervalo_inf = stats.poisson.ppf(0.025, lambda_pedidos)
intervalo_sup = stats.poisson.ppf(0.975, lambda_pedidos)
print(f"Intervalo 95%: [{intervalo_inf:.0f}, {intervalo_sup:.0f}] pedidos/hora")

print("\n ANÁLISE POR DIFERENTES PERÍODOS:")

lambda_30min = lambda_pedidos / 2
prob_30min = stats.poisson.pmf(7, lambda_30min)
print(f"P(7 pedidos em 30 min) = {prob_30min:.4f} ({prob_30min*100:.2f}%)")

lambda_2h = lambda_pedidos * 2
prob_2h = stats.poisson.pmf(30, lambda_2h)
print(f"P(30 pedidos em 2 horas) = {prob_2h:.4f} ({prob_2h*100:.2f}%)")

print("\n RECOMENDAÇÕES PARA GESTÃO:")
if prob_mais_15 > 0.3:
   print(f"Alto risco de sobrecarga ({prob_mais_15*100:.1f}%) - considere staff extra")
else:
   print(f"Risco moderado de sobrecarga ({prob_mais_15*100:.1f}%)")

print(f"Planeje capacidade para {intervalo_sup:.0f} pedidos/hora (cobertura 97.5%)")
print(f"Pedidos esperados na próxima hora: {lambda_pedidos} +- {math.sqrt(lambda_pedidos):.1f}")

print("\n RESPOSTA FINAL:")
print(f"A probabilidade da loja receber exatamente 15 pedidos em uma hora é de {prob_exato_15*100:.2f}%")

```

```python
from scipy.stats import poisson

# Define a taxa de média de pedidos por hora
lambda_ = 12

# Número de pedidos para o qual queremos calcular a probilidade
k = 15

# Número de pedidos para o qual queremos calcular 15 pedidos em um hora
probabilidade = poisson.pmf(k, lambda_)

print(f"A probabilidade de receber exatamente 15 pedidos em uma hora é de: {probabilidade * 100:.2f}%")
```

## Exercício 4

Considere uma startup de tecnologia que está testando uma nova funcionalidade no seu aplicativo. A partir de testes anteriores, a empresa estima que a probabilidade de um teste de usuário resultar em uma avaliação positiva da nova funcionalidade é de 10%. A startup está interessada em entender qual é a probabilidade de que o primeiro feedback positivo ocorra apenas no quinto teste de usuário. 

```python
import numpy as np
from scipy import stats
import math

p = 0.10
q = 1 - p
k = 5

print(" === ANÁLISE DE TESTES DE FUNCIONALIDADE === ")
print(f"Probabilidade de feedback positivo: {p*100}%")
print("Pergunta: P(primeiro sucesso no {k}° teste)")

print("DISTRIBUIÇÃO GEOMÉTRICA:")
print(f"Parâmetro p = {p}")
print(f"Número esperado de testes até o primeiro sucesso = 1/p = {1/p:.1f}")

prob_primeiro_no_5 = (q ** (k-1)) * p

print("CÁLCULO:")
print(f"P(X = {k}) = q^({k-1}) x p")
print(f"P(X = {k}) = ({q})^{k-1} x {p}")
print(f"P(X = {k}) = {q**(k-1):.6f} x {p}")
print(f"P(X = {k}) = {prob_primeiro_no_5:.6f}")
print(f"Probabilidade: {prob_primeiro_no_5*100:.2f}%")

print("\nINTERPRETAÇÃO:")
print(f" - 4 testes negativos consecutivos: {q**4:.4f} ({(q**4)*100:.2f}%)")
print(f" - 5° teste positivo: {p} ({p*100}%)")
print(f" - Combinação dos dois eventos: {prob_primeiro_no_5*100:.2f}%")

print("\n SIMULAÇÃO (10.000 experimentos):")
simulacoes = 10000
sucessos_no_5 = 0

for _ in range(simulacoes):
   for teste in range(1, 100):
      if np.random.random() < p:
         if teste == k:
            sucessos_no_5 += 1
         break
prob_simulada = sucessos_no_5 / simulacoes
print(f"Sucessos no 5° teste: {sucessos_no_5}")
print(f"Probabilidade simulada: {prob_simulada:.4f} ({prob_simulada*100:.2f})")

print("\nANÁLISES COMPLEMENTARES:")

prob_ate_5 = 1 - (q**k)
print(f"P(sucesso em até 5 testes) =  {prob_ate_5:.4f} ({prob_ate_5*100:.1f}%)")

prob_mais_10 = q ** 10
print(f"P(mais de 10 testes) = {prob_mais_10:.4f} ({prob_mais_10*100:.2f}%)")

print("\n CENÁRIOS MAIS PROVÁVEIS:")
for i in range(1, 4):
   prob_i = (q ** (i-1)) * p
   print(f"{i}° teste: {prob_i:.4f} ({prob_i*100:.2f}%)")

print(f"\nRESPOSTA:")
print(f"A probabilidade do primeiro feedback positivo ocorrer no 5° teste é de {prob_primeiro_no_5*100:.2f}%")
```

```python
from scipy.stats import geom

# Define a probabilidade de sucesso em cada tantativa
p = 0.1 # 10%

# Número de tentativas até o primeiro sucesso
k = 5

# Calcula a probabilidade de o primeiro sucesso ocorrer na 5° tentativa
probabilidade = geom.pmf(k, p)

print(f"A probabilidade de o primeiro feedback ocorrer na 5° tentativa é de {probabilidade * 100:.2f}%")
```

## Exercício 5

Imagine uma empresa de desenvolvimento de software que está trabalhando em um novo produto. Com base em projetos anteriores, a empresa sabe que a probabilidade de um teste de qualidade de software (QA) ser bem-sucedido (ou seja, o software passar no teste sem encontrar bugs) é de 80%. No entanto, para garantir a alta qualidade do produto final, a empresa define uma política de que o software deve passar por 3 testes de QA consecutivos antes de ser considerado pronto para o lançamento. A questão é: qual a probabilidade de que sejam necessários realizar 5 testes de QA para que o software passe por 3 testes consecutivos com sucesso?

```python
import numpy as np
from itertools import product

p = 0.8
q = 1 - p
sucessos_consecutivos = 3
total_testes = 5

print(" === ANÁLISE DE TESTES QA CONSECUTIVOS === ")
print(f"Probabilidade de sucesso por teste: {p*100}%")
print(f"Meta: {sucessos_consecutivos} sucessos consecutivos")
print(f"Pergunta: P(precisar de exatamente {total_testes} testes)")

print(f"\n SEQUÊNCIAS VÁLIDAS PARA {total_testes} TESTES:")

sequencias_validas = []

seq1 = "FFSSS"
seq2 = "SFSSS"

print(f"1. {seq1}: Falha -> Falha -> Sucesso -> Sucesso -> Sucesso")
print(f"2. {seq2}: Sucesso -> Falha -> Sucesso -> Sucesso -> Sucesso")

prob_FFSSS = (q**2) * (p**3)
prob_SFSSS = p * q * (p**3)

print("CÁLCULO DAS PROBABILIDADES:")
print(f"P(FFSSS) = q² x p³ = ({q})² x ({p})³ = {prob_FFSSS:.6f}")
print(f"P(SFSSS) = p x q x p³ = {p} x {q} x ({p})³ = {prob_SFSSS:.6f}")

prob_total = prob_FFSSS + prob_SFSSS

print(f"\nP(total) = P(FFSSS) + P(SFSSS)")
print(f"P(total) = {prob_FFSSS:.6f} + {prob_SFSSS:.6f}")
print(f"P(total) = {prob_total:.6f}")
print(f"Probabilidade: {prob_total*100:.2f}%")

print("\nSIMULAÇÃO (100.000 projetos): ")
simulacoes = 100000
casos_5_testes = 0

for _ in range(simulacoes):
   consecutivos = 0
   for teste in range(1, 100):
      if np.random.random() < p:
         consecutivos += 1
         if consecutivos == sucessos_consecutivos:
            if teste == total_testes:
               casos_5_testes += 1
            break
      else:
         consecutivos = 0

prob_simulada = casos_5_testes / simulacoes
print(f"Projetos que precisaram de exatamente 5 testes: {casos_5_testes}")
print(f"Probabilidade simulada: {prob_simulada:.4f} ({prob_simulada*100:.2f}%)")

print("\n ANÁLISE COMPLEMENTARES:")

prob_3_testes = p**3
print(f"P(conseguir em 3 testes) = {prob_3_testes:.4f} ({prob_3_testes*100:.1f}%)")

prob_4_testes = q * (p**3)

prob_ate_5 = prob_3_testes + prob_4_testes + prob_total
print(f"P(conseguir em até 5 testes) = {prob_ate_5:.4f} ({prob_ate_5*100:.1f}%)")

print("\n DISTRIBUIÇÃO POR NÚMERO DE TESTES:")
print(f"3 testes: {prob_3_testes*100:.1f}%")
print(f"4 testes: {prob_4_testes*100:.1f}%")
print(f"5 testes: {prob_total*100:.1f}%")
print(f"Mais de 5: {(1-prob_ate_5)*100:.1f}%")

print("\nRESPOSTA:")
print(f"A probabilidade de precisar de exatamente 5 testes para obter 3 sucessos consecutivos é de: {prob_total*100:.2f}%")
```

```python
from scipy.stats import nbinom

# Define os parâmetros da distribuição binomial negativa
r = 3 # Número de sucessos desejados (testes de QA bem-sucedido)
p = 0.8 # Probabilidade de sucesso em cada tentativa (teste de QA)

# Calcula a probabilidade de serem necessárias 5 tentativas para 3 sucessos
probabilidade = nbinom.pmf(5 - r, r, p)

print(f"A probilidade de precisar de 5 testes de QA para alcançar 3 sucessos é de: {probabilidade * 100:.2f}%")
```
